**Q3 b) Using Altair, create two linked charts side by side: a scatter plot of `trust` vs `resilience` (left) and a bar chart of mean `bii score` by `categories` (right). Add a brush selection to the scatter plot so that selecting points filters the bar chart to show only data from selected respondents. Store the combined chart as `q3b_chart`.**

<!-- TODO: generate expected output image by running this cell and saving to assets/q3b_expected.png -->

*Hint:* In Altair, create a `brush = alt.selection_interval()`. Add it to the scatter chart with `.add_params(brush)`. On the bar chart, use `.transform_filter(brush)` so it responds to selections. Concatenate with `alt.hconcat(scatter, bar)`. See [Altair selections docs](https://altair-viz.github.io/user_guide/interactions.html).

```python
# TODO: verify alt.selection_interval, add_params, transform_filter against current altair docs

### YOUR CODE HERE ###
brush = alt.selection_interval()

scatter = alt.Chart(bii).mark_point().encode(
    x=...,
    y=...,
    color=alt.condition(brush, alt.value('steelblue'), alt.value('lightgray'))
).add_params(brush)

bar = alt.Chart(bii).mark_bar().encode(
    x=...,
    y='mean(bii score):Q'
).transform_filter(brush)

q3b_chart = alt.hconcat(scatter, bar)
q3b_chart
```

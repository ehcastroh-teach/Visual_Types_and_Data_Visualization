## 3. Applying Tufte's Principles

Tufte's five principles are a practical checklist for eliminating chart junk and maximizing the data-ink ratio. This section asks you to evaluate a chart against them and produce an improved version.

**Q3 a) The cell below produces a deliberately cluttered bar chart of mean BII score by category. Identify which of Tufte's five principles it violates and which specific element in the chart causes each violation. Store your audit in `q3a_audit`.**

```python
import pandas as pd
import matplotlib.pyplot as plt

bii = pd.read_csv('./data/bii_data_w_categories.csv')
cat_means = bii.groupby('categories')['bii score'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(cat_means.index, cat_means.values,
              color=['red', 'blue', 'green', 'orange', 'purple', 'pink', 'cyan', 'yellow'],
              edgecolor='black', linewidth=2)
ax.set_facecolor('#DDDDDD')
fig.patch.set_facecolor('#BBBBBB')
ax.grid(True, color='white', linewidth=2)
ax.set_title('BII Scores by Category!!!', fontsize=24, fontweight='bold', color='red')
ax.set_xlabel('CATEGORY (nominal)', fontsize=18)
ax.set_ylabel('MEAN BII SCORE (quantitative)', fontsize=18)
ax.set_xticklabels(cat_means.index, rotation=90, fontsize=14)
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{bar.get_height():.3f}', ha='center', fontsize=10, color='black')
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
plt.tight_layout()
plt.show()
```

## 5. Generalization: Critique and Redesign a Real Chart

Apply the zero-code reasoning and Tufte principles to a domain outside the BII dataset. This section uses a chart you find or produce from the penguins dataset.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Seaborn built-in: Palmer Penguins (Horst et al. 2020)
penguins = sns.load_dataset('penguins').dropna()
```

**Q5 a) The cell below produces a chart with at least two Tufte violations. Without changing the code cell, run it and then write a critique identifying each violation and what you would fix. Store your critique in `q5a_critique`.**

```python
fig, ax = plt.subplots(figsize=(8, 6))
species_list = penguins['species'].unique()
colors = ['#FF0000', '#00FF00', '#0000FF']
for sp, col in zip(species_list, colors):
    subset = penguins[penguins['species'] == sp]
    ax.scatter(subset['bill_length_mm'], subset['flipper_length_mm'],
               c=col, s=200, edgecolors='black', linewidths=2, label=sp)
ax.set_facecolor('#EEEEEE')
ax.grid(color='white', linewidth=2)
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.set_title('Penguin Bill Length vs Flipper Length by Species!', fontsize=18, color='purple')
ax.set_xlabel('Bill Length (mm)', fontsize=14)
ax.set_ylabel('Flipper Length (mm)', fontsize=14)
ax.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

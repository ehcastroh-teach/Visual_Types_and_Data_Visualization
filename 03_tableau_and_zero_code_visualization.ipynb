## **WHEN TO USE** a **ZERO-CODE TOOL**

Most data visualization workflows begin with Python (matplotlib, seaborn, Plotly). Zero-code tools like Tableau sit beside those libraries, not above or below them. Choosing one over the other is a question of audience, iteration speed, and deployment.

**Use a zero-code tool when:**

- The primary consumer is a business stakeholder who needs to filter or explore data themselves, not just receive a static chart.
- The visualization will live in a shared dashboard refreshed on a schedule - Tableau connects natively to databases and refreshes automatically.
- Iteration speed matters more than customization: dragging a field into "Color" in Tableau is faster than writing `hue=` in seaborn and rerunning the cell.
- The final product needs to be self-service: stakeholders can answer their own follow-up questions without asking an analyst.

**Use a Python library when:**

- You need chart types or layouts that Tableau does not support natively.
- The visualization is a one-time analysis artifact that will not be shared interactively.
- You want the chart to be part of a reproducible notebook or pipeline.
- You need programmatic control over every visual element.

These are not mutually exclusive. Many teams export processed DataFrames from Python to Tableau for the presentation layer.

#### PART 1.2: BUILDING A TABLEAU DASHBOARD

<br>

A Tableau **Dashboard** is a canvas where you arrange multiple Sheets side by side or stacked. Dashboards let the audience see several related views at once and - through Actions - allow filtering one view to update another.

**Building a dashboard - step by step:**

1. Create at least two Sheets (individual charts) in the same Workbook.
2. Click the dashboard icon at the bottom of the Workbook (the icon shows a grid of squares).
3. Drag Sheets from the left panel onto the canvas. Tableau tiles them automatically.
4. Resize by dragging the borders between tiled sheets.
5. Add a **Filter Action**: Dashboard menu -> Actions -> Add Action -> Filter. Set the source sheet (the one the user clicks) and the target sheet (the one that updates). This is what makes a dashboard interactive: clicking a bar in one chart filters the second chart.

**Tableau Stories** extend dashboards into a narrative: each Story Point is a Dashboard or Sheet with an annotation. Use Stories when you need to walk an audience through a sequence of findings, not just give them a self-service tool.

___

**Note:** Tableau Online does not support all desktop features. Calculated fields, LOD (Level of Detail) expressions, and most chart types are fully available. TabPy integration (Part 1.3) requires a hosted TabPy server.

#### PART 1.3: TABLEAU AND PYTHON (TABPY)

<br>

TabPy (Tableau Python Server) is an open-source server that lets Tableau call Python functions as calculated fields. This is useful when your analysis requires something Python does well but Tableau cannot do natively: running a scikit-learn model, computing a custom aggregation, or calling an external API.

**How it works:**

1. You run a TabPy server locally (or on a remote host).
2. Tableau connects to it under Help -> Settings and Performance -> Manage Analytics Extension Connection.
3. In a calculated field, you write `SCRIPT_REAL("...", ...)` - Tableau sends the expression and the data to the TabPy server, Python evaluates it, and returns the result to Tableau as a column.

**When to use TabPy:**

- You need a machine learning prediction column that Tableau cannot compute.
- You want to call a custom Python aggregation that Tableau's built-in LOD expressions cannot express.
- You need to enrich Tableau data with results from an external API.

**When not to use TabPy:**

- For anything Tableau's calculated fields can already do. TabPy adds latency and a server dependency.
- For production deployments where uptime matters - every Tableau user who opens the workbook triggers a Python call, which the TabPy server must handle.

___

**Note:** TabPy requires a running server. For local development: `pip install tabpy-server && tabpy`. The Tableau Online connection to a local TabPy server requires Tableau Bridge or a publicly reachable host. See: https://github.com/tableau/TabPy

___

## APPENDIX I: Zero-Coding Alternatives and Dashboards


> * **BOKEH Visualization Library** A slightly different flavor to Plotly, just as good if not better according to some.<br>
>https://bokeh.pydata.org

> * **BQPLOT Visualization Library** Developed by Bloomberg media, this library can be quickly learned and offers many of the benefits of more challenging libraries.<br>
>https://pypi.org/project/bqplot/

> * **ALTAIR Visualization Library** The backbone to many declerative plotting engines, it is always worth playing with.<br>
>https://altair-viz.github.io/

> * **VEGA Visualization Library** A monster of a visualization engine when properly used.  Not easy to pick up, but once learned it can keep up with the best.<br>
>https://vega.github.io/vega/

> * **K3D Jupyter Widget** Part of a new breed of visualization libraries trying to make dynamic and interactive visualizations natively work in jupyter.<br>
>https://k3d-jupyter.org/ &nbsp; and &nbsp; https://jupyter.org/widgets

> * **DASH: Dashboard Based Dynamic Plotting** Works seamlessly with plotly, and enables the user to use simple click and drag mechanics to fully customize your dashboards.<br>
>https://dash-gallery.plotly.host/Portal/

> * **TABLEAU: Web Based Visualization Environment** Basically the standard for enterprise-focused visualizations.
> https://public.tableau.com/s/

> * **INFOGRAM: Web Based Comprehensive Visualization and Inforgraphic Environment** A one stop shop for all of your visualization needs.<br>
>https://infogram.com/

> * **FLOURISH: Web Based Visualization Environment** Are you a spreadsheet junkie, but hate the plotting options available by default? Boy, you should check out Flourish.<br>
>https://flourish.studio/

> * **DATAWRAPPER: Web Based Visualization and Map Creation Environment** more htan just a nieche service, Datawrapper offers some powerful capabilities through its web based service.<br>
